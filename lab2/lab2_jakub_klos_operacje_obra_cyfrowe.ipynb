## Operacja LUT

**Operacja LUT** polega na przekształcaniu wartości poszczególnych pikseli obrazu przy użyciu z góry przygotowanych tabel przekodowań (tabel korekcji).

W przetwarzaniu obrazów najczęściej wykorzystuje się następujące funkcje:
- typu kwadratowa, pierwiastek kwadratowy
- typu logarytm, odwrócony logarytm
- typu wykładnicza,
- inne (np. piłokształtna).

Aby lepiej zobaczyć w jaki sposób działają różne przekodowania LUT, skonstruujemy funkcję, która jako argumenty pobierać będzie obrazek oryginalny oraz tablicę przekodowania, a następnie na wspólnym rysunku będzie wyświetlać: funkcję, obraz wejściowy oraz wynik przekodowania.

Przypomnienie składni tworzenia funkcji w pythonie:
```{python}
  def nazwa_funkcji(para, metry):
  	# cialo funkcji
```

  - Stwórz nową funkcję i nazwij ją LUT.
  - Funkcja powinna przyjmować dwa parametry: obraz oraz tablicę przekodowania.
  - W ciele funkcji wykonaj przekodowanie LUT, podobnie jak wcześniej w przykładzie.
  - Funkcja powinna wyświetlić wykres składający się z 3 umieszczonych obok siebie pól: wykres przekodowania, obraz oryginalny oraz obraz przekształcony.
    Każdy z wykresów powinien być podpisany.
    _(W razie problemów można przypomnieć sobie te zagadnienia z laboratorium wprowadzającego)_
  - Jeśli wykres przekodowania jest zbyt rozciągnięty, można go wyrównać, np. `ax2.set_aspect('equal')`.

````python
import cv2
import numpy as np
import matplotlib.pyplot as plt
import lut


def LUT(image, lut_table):
    """
    Funkcja stosująca operację LUT i wizualizująca wyniki.

    Parametry:
    - image: Obraz w skali szarości
    - lut_table: Tablica przekodowania LUT
    """
    transformed_image = cv2.LUT(image, lut_table)

    x = np.arange(256)  
    y = lut_table[x]  

    fig, axes = plt.subplots(1, 3, figsize=(15, 5))

    axes[0].plot(x, y, color="blue", label="Przekodowanie LUT")
    axes[0].set_title("Funkcja przekodowania LUT")
    axes[0].set_xlabel("Oryginalna wartość piksela")
    axes[0].set_ylabel("Nowa wartość piksela")
    axes[0].grid()
    axes[0].set_aspect('equal')

    axes[1].imshow(image, cmap='gray')
    axes[1].set_title("Oryginalny obraz")
    axes[1].axis("off")

    axes[2].imshow(transformed_image, cmap='gray')
    axes[2].set_title("Obraz po LUT")
    axes[2].axis("off")

    plt.show()

LUT(image, lut.log)  
````

Wywołaj przygotowaną funkcję dla każdego z przekształceń.
W wyniku powinno powstać 7 wykresów.

````python
lut_functions = [
    ("Logarytmiczna", lut.log),
    ("Odwrotna logarytmiczna", lut.odwlog),
    ("Pierwiastkowa", lut.pierwiastkowa),
    ("Wykładnicza", lut.wykladnicza),
    ("Kwadratowa", lut.kwadratowa),
    ("Piłokształtna", lut.pila),
    ("Odwrotna", lut.odwrotna)
]

for name, lut_func in lut_functions:
    print(f"Stosowanie LUT: {name}")
    LUT(image, lut_func)
````

## Operacja arytmetyczne

### Dodawanie

Wczytaj dwa obrazy _lena.bmp_ i _jet.bmp_ i wyświetl je.

````python
import cv2
import os
import matplotlib.pyplot as plt

def download_and_load_image(filename):
    """ Pobiera obraz za pomocą `wget`, jeśli nie istnieje, i wczytuje go jako obraz w skali szarości. """
    url = f"https://raw.githubusercontent.com/vision-agh/poc_sw/master/02_Point/{filename}"
    
    if not os.path.exists(filename):
        !wget {url} -O {filename}

    image = cv2.imread(filename, cv2.IMREAD_GRAYSCALE)

    if image is None:
        raise ValueError(f"Nie udało się wczytać obrazu {filename}. Sprawdź, czy plik istnieje i jest poprawny.")
    
    return image

image_lena = download_and_load_image("lena.bmp")
image_jet = download_and_load_image("jet.bmp")

plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
plt.imshow(image_lena, cmap='gray')
plt.title("Lena.bmp")
plt.xticks([]), plt.yticks([])

plt.subplot(1, 2, 2)
plt.imshow(image_jet, cmap='gray')
plt.title("Jet.bmp")
plt.xticks([]), plt.yticks([])

plt.tight_layout()
plt.show()
````

Dodaj obrazy _lena_ i _jet_, wykorzystaj funkcję `cv2.add`.
Uzyskany wynik wyświetl.

````python
added_image = cv2.add(image_lena, image_jet)

plt.figure(figsize=(6, 6))
plt.imshow(added_image, cmap='gray')
plt.title("Dodane obrazy (lena.bmp + jet.bmp)")
plt.xticks([]), plt.yticks([])
plt.show()
````

Czy wynik sumowania jest satysfakcjonujący?
Co może niekorzystnie wpływać na rezultat operacji?
Spróbuj wykonać dodawanie ponownie wykorzystując typ uint16 (`jet.astype('uint16')`).
Wynikowy obraz należy przeskalować do zakresu 0-255, zamienić na typ uint8 i wyświetlić. _
Uwaga: operacja ta jest użyteczna w przypadku, gdy dane do wyświetlenia wykraczają poza zakres 0-255, w przeciwnym przypadku jej wykorzystanie może zniekształcić wyniki._

````python
import numpy as np
import matplotlib.pyplot as plt

lena_uint16 = image_lena.astype('uint16')
jet_uint16 = image_jet.astype('uint16')

added_uint16 = lena_uint16 + jet_uint16

max_val = np.max(added_uint16)
print(f"Maksymalna wartość piksela po dodaniu: {max_val}")

added_normalized = ((added_uint16 / max_val) * 255).astype('uint8')

added_clipped = np.clip(added_uint16, 0, 255).astype('uint8')

plt.figure(figsize=(15, 5))

plt.subplot(1, 3, 1)
plt.imshow(image_lena, cmap='gray')
plt.title('Oryginalny Lena')
plt.xticks([]), plt.yticks([])

plt.subplot(1, 3, 2)
plt.imshow(image_jet, cmap='gray')
plt.title('Oryginalny Jet')
plt.xticks([]), plt.yticks([])

plt.figure(figsize=(15, 5))

plt.subplot(1, 3, 1)
plt.imshow(added_clipped, cmap='gray')
plt.title('Dodawanie z obcięciem (uint8)')
plt.xticks([]), plt.yticks([])

plt.subplot(1, 3, 2)
plt.imshow(added_normalized, cmap='gray')
plt.title('Dodawanie z normalizacją (uint8)')
plt.xticks([]), plt.yticks([])

plt.tight_layout()
plt.show()
````

### Odejmowanie

Wykorzystując funkcję `cv2.subtract` odejmij obrazy _lena_ i _jet_.

````python
odejmowanie_image = cv2.subtract(image_lena, image_jet)

plt.figure(figsize=(6, 6))
plt.imshow(odejmowanie_image, cmap='gray')
plt.title("Odejmowanie obrazów (lena.bmp - jet.bmp)")
plt.xticks([]), plt.yticks([])
plt.show()

odejmowanie_image2 = cv2.subtract(image_jet, image_lena)

plt.figure(figsize=(6, 6))
plt.imshow(odejmowanie_image2, cmap='gray')
plt.title("Odejmowanie obrazow (jet.bmp - lena.bmp)")
plt.xticks([]), plt.yticks([])
plt.show()

````

Czy wynik odejmowania jest satysfakcjonujący?
Co może niekorzystnie wpływać na rezultat operacji?
Często zamiast zwykłego odejmowania wykorzystuje się operację wartość bezwzględna z różnicy (pozwala to między innymi uniknąć pokazanych powyżej problemów).
Zamień typ obrazów _lena_ i _jet_ z uint8 na **int16**, odejmij je od siebie, a następnie wykorzystując funkcję `np.abs` wykonaj operację wartość bezwzględna z różnicy.
Wyświetl ten obraz.
Zastanów się, dlaczego ta zmiana poprawia wynik odejmowania?

````python
import cv2
import numpy as np
import matplotlib.pyplot as plt

lena_int16 = image_lena.astype('int16')
jet_int16 = image_jet.astype('int16')

subtracted_image = lena_int16 - jet_int16

abs_difference = np.abs(subtracted_image)

abs_difference_scaled = (abs_difference / np.max(abs_difference) * 255).astype('uint8')

plt.figure(figsize=(6, 6))
plt.imshow(abs_difference_scaled, cmap='gray')
plt.title("Odejmowanie |Lena - Jet| (wartość bezwzględna)")
plt.xticks([]), plt.yticks([])
plt.show()
````

Na poszczególnych punktach obrazu (najczęściej binarnego $-$ czyli składającego się z dwóch kolorów: czarnego i białego) można wykonywać operacje logiczne: NOT, AND, OR, XOR itp.
Wczytaj dwa obrazy: _kolo.bmp_ i _kwadrat.bmp_ (https://raw.githubusercontent.com/vision-agh/poc_sw/master/02_Point/kwadrat.bmp), następnie wyświetl je.

Na wczytanych obrazach wykonaj wybrane operacje logiczne: NOT (operator `~`), AND (`&`), OR (`|`), XOR (`^`).
Operator `~` jest jednoargumentowy, wszystkie pozostałe są dwuargumentowe.
Alternatywnym sposobem jest wykorzystanie funkcji z biblioteki opencv: `cv2.bitwise_not`, `cv2.bitwise_and`, `cv2.bitwise_or`, `cv2.bitwise_xor`.

````python
import cv2
import os
import numpy as np
import matplotlib.pyplot as plt

def check_and_download_image(filename):
    if not os.path.exists(filename):
        print(f"Pobieranie obrazu {filename}...")
        url = f"https://raw.githubusercontent.com/vision-agh/poc_sw/master/02_Point/{filename}"
        !wget {url} -O {filename}
    return cv2.imread(filename, cv2.IMREAD_GRAYSCALE)

kolo = check_and_download_image("kolo.bmp")
kwadrat = check_and_download_image("kwadrat.bmp")
lena = check_and_download_image("lena.bmp")
jet = check_and_download_image("jet.bmp")


def display_logical_operations(img1, img2, title1, title2):
    not_img1 = ~img1
    not_img2 = ~img2
    and_result = img1 & img2
    or_result = img1 | img2
    xor_result = img1 ^ img2

    f, axes = plt.subplots(2, 4, figsize=(16, 8))
    f.suptitle(f'Operacje logiczne (operatory): {title1} i {title2}', fontsize=16)

    axes[0, 0].imshow(img1, cmap='gray')
    axes[0, 0].set_title(f'Oryginalny {title1}')
    axes[0, 1].imshow(img2, cmap='gray')
    axes[0, 1].set_title(f'Oryginalny {title2}')
    axes[0, 2].imshow(not_img1, cmap='gray')
    axes[0, 2].set_title(f'NOT {title1}')
    axes[0, 3].imshow(not_img2, cmap='gray')
    axes[0, 3].set_title(f'NOT {title2}')
    axes[1, 0].imshow(and_result, cmap='gray')
    axes[1, 0].set_title(f'AND ({title1} & {title2})')
    axes[1, 1].imshow(or_result, cmap='gray')
    axes[1, 1].set_title(f'OR ({title1} | {title2})')
    axes[1, 2].imshow(xor_result, cmap='gray')
    axes[1, 2].set_title(f'XOR ({title1} ^ {title2})')
    axes[1, 3].axis('off')

    for ax in axes.flatten():
        ax.set_xticks([])
        ax.set_yticks([])

    plt.tight_layout()
    plt.subplots_adjust(top=0.9)
    plt.show()

def display_logical_operations_cv2(img1, img2, title1, title2):
    not_img1 = cv2.bitwise_not(img1)
    not_img2 = cv2.bitwise_not(img2)
    and_result = cv2.bitwise_and(img1, img2)
    or_result = cv2.bitwise_or(img1, img2)
    xor_result = cv2.bitwise_xor(img1, img2)

    f, axes = plt.subplots(2, 4, figsize=(16, 8))
    f.suptitle(f'Operacje logiczne (OpenCV): {title1} i {title2}', fontsize=16)

    axes[0, 0].imshow(img1, cmap='gray')
    axes[0, 0].set_title(f'Oryginalny {title1}')
    axes[0, 1].imshow(img2, cmap='gray')
    axes[0, 1].set_title(f'Oryginalny {title2}')
    axes[0, 2].imshow(not_img1, cmap='gray')
    axes[0, 2].set_title(f'NOT {title1}')
    axes[0, 3].imshow(not_img2, cmap='gray')
    axes[0, 3].set_title(f'NOT {title2}')
    axes[1, 0].imshow(and_result, cmap='gray')
    axes[1, 0].set_title(f'AND ({title1} & {title2})')
    axes[1, 1].imshow(or_result, cmap='gray')
    axes[1, 1].set_title(f'OR ({title1} | {title2})')
    axes[1, 2].imshow(xor_result, cmap='gray')
    axes[1, 2].set_title(f'XOR ({title1} ^ {title2})')
    axes[1, 3].axis('off')

    for ax in axes.flatten():
        ax.set_xticks([])
        ax.set_yticks([])

    plt.tight_layout()
    plt.subplots_adjust(top=0.9)
    plt.show()

display_logical_operations_cv2(kolo, kwadrat, "Koło", "Kwadrat")

display_logical_operations_cv2(lena, jet, "Lena", "Jet")
````
